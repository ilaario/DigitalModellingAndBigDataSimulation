## Conditions and loops

```python
# Exercise 11
# Loop over these waiting times (minutes). Print "soon" if 5 or less, otherwise "wait".
times = [3, 12, 5, 20, 8]

for waiting_time in times:
    if waiting_time <= 5:
        print("soon")
    else:
        print("wait")
```

```python
# Exercise 13
# Use enumerate to print each station with its number starting at 1.
stations = ["Nord", "Mitte", "Sued"]

for position, station in enumerate(stations, start=1):
    print(position, station)
```

## Functions

```python
# Exercise 14
# Write a function minutes_per_km(total_minutes, km) that returns pace per km.
# Print it for a 45 minute, 9 km ride.

def minutes_per_km(total_minutes, km):
    return total_minutes / km

print(minutes_per_km(45, 9))
```

```python
# Exercise 15
# Write a function fare(zones, price_per_zone=2.4) that returns the total fare.
# Test it for 3 zones with the default, and for 3 zones at 2.0 per zone.

def fare(zones, price_per_zone=2.4):
    return zones * price_per_zone

print(fare(3))
print(fare(3, 2.0))
```

## Comprehension and classes

```python
# Exercise 16
# From the numbers 1 to 12, build a list of only the ones divisible by 3.

multiples_of_three = [number for number in range(1, 13) if number % 3 == 0]
print(multiples_of_three)
```

```python
# Exercise 17
# Define a class Trip with distance_km and minutes set in __init__, and a method
# speed_kmh() that returns distance_km divided by (minutes / 60).
# Make a Trip of 15 km in 30 minutes and print the speed.

class Trip:
    def __init__(self, distance_km, minutes):
        self.distance_km = distance_km
        self.minutes = minutes

    def speed_kmh(self):
        return self.distance_km / (self.minutes / 60)

trip = Trip(15, 30)
print(trip.speed_kmh())
```

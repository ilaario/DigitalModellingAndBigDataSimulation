## 9. Loops and conditions

```python
# Exercise 19: loop over these distances. For each, print "far" if it is above 100,
# otherwise print "near".
distances = [45, 80, 120, 30]

for distance in distances:
    if distance > 100:
        print("far")
    else:
        print("near")
```

```python
# Exercise 20: using range, print the even numbers from 0 to 10.
# Hint: range(start, stop, step), or use an if with %

for number in range(0, 11, 2):
    print(number)
```

```python
# Exercise 21: use enumerate to print each city with its position starting at 1,
# like "1: Vienna", "2: Graz".
cities = ["Vienna", "Graz", "Linz"]

for position, city in enumerate(cities, start=1):
    print(f"{position}: {city}")
```

```python
# Exercise 22: start at 100. Using a while loop, keep halving the value and printing it,
# stopping once it drops below 1.

value = 100
while value >= 1:
    value = value / 2
    print(value)
```

## 10. Functions

```python
# Exercise 23: write a function kmh_to_ms(kmh) that returns the speed in metres per second
# (divide by 3.6). Print the result for 100 km/h.

def kmh_to_ms(kmh):
    return kmh / 3.6

print(kmh_to_ms(100))
```

```python
# Exercise 24: write a function label_speed(speed, limit=100) that returns "over" if the
# speed is above the limit, otherwise "ok". Test it with 120, and with 90.

def label_speed(speed, limit=100):
    if speed > limit:
        return "over"
    return "ok"

print(label_speed(120))
print(label_speed(90))
```

## 11. List comprehension and a common trap

```python
# Exercise 25: from the numbers 1 to 10, build a list of the squares of only the even numbers.
# Hint: range(1, 11), a condition with %

squares = [number ** 2 for number in range(1, 11) if number % 2 == 0]
print(squares)
```

```python
# Exercise 26: -999 means "missing". Build a NEW list that keeps only the real values.
readings = [21.0, 22.5, -999, 23.1, -999, 24.0, 20.8]

clean = [reading for reading in readings if reading != -999]
print(clean)
```

## 12. Classes

```python
# Exercise 27: define a class Car with brand and range_km set in __init__,
# and a method is_long_range() that returns True when range_km is 400 or more.
# Create a Car with range 520 and print the result of is_long_range().

class Car:
    def __init__(self, brand, range_km):
        self.brand = brand
        self.range_km = range_km

    def is_long_range(self):
        return self.range_km >= 400

car = Car("Tesla", 520)
print(car.is_long_range())
```
